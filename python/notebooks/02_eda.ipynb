```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

ROOT = Path('../../')
df = pd.read_csv(ROOT / 'dataset/cleaned/cleaned_vehicle_sales.csv', nrows=80000)
sns.histplot(df['sellingprice'], bins=50, kde=True)
plt.title('Price Distribution'); plt.show()
```

```python
sns.scatterplot(data=df.sample(5000), x='odometer', y='sellingprice', alpha=0.3)
plt.title('Odometer vs Price'); plt.show()
```

```python
numeric = df[['year','odometer','condition','vehicle_age','mmr','sellingprice']].corr()
sns.heatmap(numeric, annot=True, fmt='.2f'); plt.show()
```

```python
top = df['make'].value_counts().head(8).index
sns.boxplot(data=df[df['make'].isin(top)], x='make', y='sellingprice')
plt.xticks(rotation=45); plt.show()
```
