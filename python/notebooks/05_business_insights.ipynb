# 05 — Business Insights
Revenue, MMR premium, seasonal & seller scorecards.

```python
import pandas as pd
df = pd.read_csv('../../dataset/cleaned/cleaned_vehicle_sales.csv', nrows=150000)
(df.groupby('make')['sellingprice'].sum().sort_values(ascending=False).head(10) / 1e6).round(2)
```

```python
df.groupby('sale_quarter')['sellingprice'].mean().round(2)
```

```python
underpriced = df[df['sellingprice'] < 0.9 * df['mmr']]
print('Units:', len(underpriced))
print('Potential profit @ MMR:', (underpriced['mmr'] - underpriced['sellingprice']).sum())
```
